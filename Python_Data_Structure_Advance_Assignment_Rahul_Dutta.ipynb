**Q1. Write a code to reverse a string?**


**Ans1.** Here’s a simple Python code to reverse a string using slicing, which is one of the most efficient ways in Python:

### Code:

````python
# Function to reverse a string
def reverse_string(input_string):
    # Using slicing to reverse the string
    reversed_string = input_string[::-1]
    return reversed_string

# Test the function
user_input = input("Enter a string to reverse: ")
result = reverse_string(user_input)
print(f"Reversed string: {result}")
````

### How it works:
1. **Slicing**: The `[::-1]` slice starts from the end of the string and moves backward, effectively reversing it.
2. **Function**: The `reverse_string` function makes the code reusable.

### Example Input/Output:

````python
Enter a string to reverse: hello
Reversed string: olleh
````

**Q2. Write a code to count the number of vowels in a string?**

**Ans2.** Here’s a Python code to count the number of vowels in a string:

### Code:




````python
# Function to count vowels in a string
def count_vowels(input_string):
    # Define the vowels
    vowels = "aeiouAEIOU"
    # Initialize a counter
    count = 0

    # Iterate through each character in the string
    for char in input_string:
        # Check if the character is a vowel
        if char in vowels:
            count += 1

    return count

# Test the function
user_input = input("Enter a string: ")
vowel_count = count_vowels(user_input)
print(f"Number of vowels in the string: {vowel_count}")
````

### How it works:
1. **Vowels Set**: The string `vowels` contains all uppercase and lowercase vowels.
2. **Loop**: The `for` loop goes through each character in the input string.
3. **Check**: For each character, it checks if it’s in the `vowels` string.
4. **Count**: If it's a vowel, the counter increases by 1.

**Q3. Write a code to check if a given string is a palindrome or not?**


**Ans3.** Here’s a Python code to check if a string is a palindrome:

### Code:

````python
# Function to check if a string is a palindrome
def is_palindrome(input_string):
    # Remove spaces and convert to lowercase for consistent comparison
    cleaned_string = input_string.replace(" ", "").lower()
    # Check if the string is the same when reversed
    return cleaned_string == cleaned_string[::-1]

# Test the function
user_input = input("Enter a string to check if it's a palindrome: ")
if is_palindrome(user_input):
    print("The string is a palindrome!")
else:
    print("The string is not a palindrome.")
````


### How it works:
1. **Cleaning the String**:
   - Spaces are removed using `replace(" ", "")`.
   - The string is converted to lowercase using `.lower()` for case-insensitive comparison.
2. **Palindrome Check**:
   - The string is compared with its reverse (`[::-1]`).
3. **Output**: Based on the comparison, it prints whether the string is a palindrome.

### Example Input/Output:



````python
Enter a string to check if it's a palindrome: Madam
The string is a palindrome!

Enter a string to check if it's a palindrome: Hello
The string is not a palindrome.
````

**Q5.Write a code to find all occurrences of a given substring within another string.**

**Ans5.** Here's a Python program to find all occurrences of a given substring within another string:

````python
def find_occurrences(text, substring):
    positions = []
    start = 0

    while start < len(text):
        pos = text.find(substring, start)
        if pos == -1:
            break
        positions.append(pos)
        start = pos + 1  # Move to the next character after the found substring

    return positions

# Example usage
text = input("Enter the main string: ")
substring = input("Enter the substring to find: ")

positions = find_occurrences(text, substring)

if positions:
    print(f"Substring found at positions: {positions}")
else:
    print("Substring not found!")
````

### Explanation:
- The function `find_occurrences(text, substring)` uses a `while` loop to search for the substring using `.find()`.
- It keeps track of found positions in a list and updates `start` to continue searching after the last found position.
- If found, it prints all positions; otherwise, it prints "Substring not found!"



**Q6.Write a code to perform basic string compression using the counts of repeated characters.**

**Ans6.** Here's a Python program to perform basic string compression using counts of repeated characters:

````python
def compress_string(s):
    compressed = []
    count = 1

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            compressed.append(s[i - 1] + str(count))
            count = 1

    compressed.append(s[-1] + str(count))  # Add the last character and its count
    compressed_string = "".join(compressed)

    return compressed_string if len(compressed_string) < len(s) else s  # Return original if compression isn't smaller

# Example usage
input_string = input("Enter a string: ")
compressed = compress_string(input_string)
print("Compressed String:", compressed)
````

  
### Explanation:
- We iterate through the string and count consecutive repeating characters.
- Whenever a new character appears, we append the previous character and its count to a list.
- At the end, we join the list into a compressed string.
- If the compressed string is not shorter than the original, we return the original string.

#### Example:
```
Input: "aabcccccaaa"
Output: "a2b1c5a3"
```

**Q14. Write a code to reverse a list in-place without using any built-in reverse functions.**


**Ans14.** Here's a Python program to reverse a list **in-place** without using any built-in `reverse()` function:

````python
def reverse_list(lst):
    left, right = 0, len(lst) - 1  # Pointers to the start and end of the list
    while left < right:
        lst[left], lst[right] = lst[right], lst[left]  # Swap elements
        left += 1
        right -= 1

# Example usage
input_list = list(map(int, input("Enter list elements separated by spaces: ").split()))

reverse_list(input_list)
print("Reversed List:", input_list)
````

### Explanation:
- We use **two pointers**: `left` (starting index) and `right` (last index).
- We **swap** elements at `left` and `right` positions.
- **Move** `left` forward and `right` backward until they meet.



````python
Input:
  List: 1 2 3 4 5
Output:
  Reversed List: [5, 4, 3, 2, 1]
````

**Q15.Implement a code to find and remove duplicates from a list while preserving the original order of elements .**

**Ans15.** Here's a Python program to **find and remove duplicates** from a list while **preserving the original order** of elements:  

````python
def remove_duplicates(lst):
    seen = set()  # Set to track seen elements
    unique_list = []  # List to store unique elements

    for num in lst:
        if num not in seen:
            unique_list.append(num)  # Add only if not already seen
            seen.add(num)  # Mark as seen

    return unique_list

# Example usage
input_list = list(map(int, input("Enter list elements separated by spaces: ").split()))

result = remove_duplicates(input_list)
print("List after removing duplicates:", result)
````

### Explanation:
- We use a **set (`seen`)** to track elements that have already appeared.
- We iterate through the list and **add elements only if they haven't been seen before**.
- This ensures **duplicates are removed** while maintaining the **original order**.




**Example:**

````python
Input: 4 2 4 3 2 1 4 1 3 3
Output: List after removing duplicates: [4, 2, 3, 1]
````

**Q17. Write a code to merge two sorted lists into a single sorted list.**

**Ans17.** Here's a Python program to **merge two sorted lists** into a **single sorted list** efficiently:

````python
def merge_sorted_lists(lst1, lst2):
    merged_list = []
    i, j = 0, 0  # Pointers for both lists

    # Merge elements from both lists in sorted order
    while i < len(lst1) and j < len(lst2):
        if lst1[i] < lst2[j]:
            merged_list.append(lst1[i])
            i += 1
        else:
            merged_list.append(lst2[j])
            j += 1

    # Add remaining elements if any
    merged_list.extend(lst1[i:])
    merged_list.extend(lst2[j:])

    return merged_list

# Example usage
list1 = list(map(int, input("Enter first sorted list (space-separated): ").split()))
list2 = list(map(int, input("Enter second sorted list (space-separated): ").split()))

result = merge_sorted_lists(list1, list2)
print("Merged Sorted List:", result)
````

### Explanation:
1. **Use two pointers** (`i` and `j`) to traverse both lists.
2. Compare elements and **add the smaller one** to the `merged_list`.
3. If one list is exhausted, **append the remaining elements** from the other list.





**Example:**

````python
Input:
  First list: 1 3 5 7
  Second list: 2 4 6 8
Output:
  Merged Sorted List: [1, 2, 3, 4, 5, 6, 7, 8]
````

**Q20. Write a code to shuffle a given list randomly without using any built-in shuffle functions.**

**Ans20.** Here's a Python program to **shuffle a list randomly** without using built-in `shuffle()` functions:

````python
import random

def custom_shuffle(lst):
    n = len(lst)
    for i in range(n - 1, 0, -1):  # Start from the last index to the first
        j = random.randint(0, i)  # Pick a random index from 0 to i
        lst[i], lst[j] = lst[j], lst[i]  # Swap elements

# Example usage
input_list = list(map(int, input("Enter list elements separated by spaces: ").split()))

custom_shuffle(input_list)
print("Shuffled List:", input_list)
````

### Explanation:
- We use the **Fisher-Yates (Knuth) Shuffle Algorithm**, which efficiently shuffles the list in **O(n) time**.
- We start from the **last element** and swap it with a **random element** before it.
- The **random index (`j`)** ensures uniform randomness.



**Example:**

````python
Input: 1 2 3 4 5
Output: Shuffled List: [3, 5, 1, 2, 4]  (Output will vary)
````

**Q32. Write a code that takes two dictionaries as input and merges them into a single dictionary. If there are common keys, the values should be added together.**

**Ans32.** Here's a Python program that takes **two dictionaries as input** and merges them into a **single dictionary**, adding values for common keys:

````python
def merge_dictionaries(dict1, dict2):
    merged_dict = dict1.copy()  # Copy the first dictionary to avoid modifying it
    for key, value in dict2.items():
        merged_dict[key] = merged_dict.get(key, 0) + value  # Add values for common keys
    return merged_dict

# Taking user input for two dictionaries
dict1 = eval(input("Enter the first dictionary (e.g., {'a': 2, 'b': 3}): "))
dict2 = eval(input("Enter the second dictionary (e.g., {'b': 4, 'c': 5}): "))

# Merging dictionaries
result = merge_dictionaries(dict1, dict2)

# Printing the merged dictionary
print("Merged Dictionary:", result)
````

### Explanation:
- We **copy `dict1`** to avoid modifying the original.
- We **iterate through `dict2`**:
  - If a key exists in both, we **add their values**.
  - If a key is only in `dict2`, it is **added to `dict1`**.
- **`eval(input())`** allows entering dictionaries in standard Python format.




**Example:**

````python
Input:
  First dictionary: {'a': 2, 'b': 3}
  Second dictionary: {'b': 4, 'c': 5}
Output:
  Merged Dictionary: {'a': 2, 'b': 7, 'c': 5}
````

**Q33. Write a code to access a value in a nested dictionary. The function should take the dictionary and a list of keys as input, and return the corresponding value. If any of the keys do not exist in the dictionary, the function should return None .**

**Ans33.** Here's a Python program that **accesses a value in a nested dictionary** using a **list of keys**:

````python
def get_nested_value(nested_dict, keys):
    current = nested_dict
    for key in keys:
        if isinstance(current, dict) and key in current:
            current = current[key]  # Move deeper into the dictionary
        else:
            return None  # Key not found
    return current  # Return the final value

# Example nested dictionary
nested_dict = {
    'a': {
        'b': {
            'c': 42
        }
    }
}

# Taking user input for keys as a comma-separated list
keys = input("Enter the keys separated by commas (e.g., a,b,c): ").split(',')

# Getting the value from the nested dictionary
result = get_nested_value(nested_dict, keys)

# Printing the result
print("Value:", result)
````

### Explanation:
- We **traverse the dictionary** using the list of keys.
- If a key exists, we move deeper; if a key **doesn't exist**, return `None`.
- **User input** is taken as a comma-separated string and converted into a **list of keys**.




**Example:**

````python
Input:
  Keys: a,b,c
Output:
  Value: 42

Input:
  Keys: a,b,d
Output:
  Value: None
````

**Q34. Write a code that takes a dictionary as input and returns a sorted version of it based on the values. You can choose whether to sort in ascending or descending order.**

**Ans34.** Here's a Python program that takes a **dictionary as input** and returns a **sorted version based on values**, allowing the user to choose **ascending or descending order**:

````python
def sort_dict_by_value(input_dict, ascending=True):
    return dict(sorted(input_dict.items(), key=lambda item: item[1], reverse=not ascending))

# Taking user input for the dictionary
input_dict = eval(input("Enter a dictionary (e.g., {'a': 3, 'b': 1, 'c': 2}): "))

# Taking user choice for sorting order
order = input("Sort in ascending order? (yes/no): ").strip().lower()
ascending = order == "yes"

# Sorting dictionary
sorted_dict = sort_dict_by_value(input_dict, ascending)

# Printing the sorted dictionary
print("Sorted Dictionary:", sorted_dict)
````

### Explanation:
- **`sorted(input_dict.items(), key=lambda item: item[1])`** sorts dictionary items **by values**.
- **`reverse=not ascending`** ensures sorting is:
  - **Ascending** if `ascending=True`
  - **Descending** if `ascending=False`
- The user **enters a dictionary** and chooses sorting order (`yes` for ascending).




**Example:**

````python
Input:
  Dictionary: {'a': 3, 'b': 1, 'c': 2}
  Sort in ascending order? yes
Output:
  Sorted Dictionary: {'b': 1, 'c': 2, 'a': 3}

Input:
  Dictionary: {'a': 3, 'b': 1, 'c': 2}
  Sort in ascending order? no
Output:
  Sorted Dictionary: {'a': 3, 'c': 2, 'b': 1}
````

**Q35. Write a code that inverts a dictionary, swapping keys and values. Ensure that the inverted dictionary correctly handles cases where multiple keys have the same value by storing the keys as a list in the inverted dictionary.**

**Ans35.** Here’s a Python program that **inverts a dictionary**, swapping keys and values while **handling duplicate values** by storing multiple keys in a list:

````python
def invert_dictionary(input_dict):
    inverted_dict = {}  # Dictionary to store the inverted result
    for key, value in input_dict.items():
        if value in inverted_dict:
            inverted_dict[value].append(key)  # Append key if value already exists
        else:
            inverted_dict[value] = [key]  # Create a new list for this value
    return inverted_dict

# Taking user input for the dictionary
input_dict = eval(input("Enter a dictionary (e.g., {'a': 1, 'b': 2, 'c': 1}): "))

# Inverting the dictionary
inverted_dict = invert_dictionary(input_dict)

# Printing the inverted dictionary
print("Inverted Dictionary:", inverted_dict)
````

### Explanation:
- **Swaps keys and values** while ensuring that if **multiple keys** have the same value, they are **stored in a list**.
- **Checks if a value already exists** in `inverted_dict`:
  - If **yes** → **Appends** the key to the list.
  - If **no** → **Creates a new list** with the key.
- Uses **`eval(input())`** to allow user input in Python dictionary format.




**Example:**

````python
Input:
  {'a': 1, 'b': 2, 'c': 1, 'd': 3}
Output:
  Inverted Dictionary: {1: ['a', 'c'], 2: ['b'], 3: ['d']}
````
